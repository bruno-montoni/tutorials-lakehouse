### Core Concepts ###
```fastf1``` organizes Formula 1 data around an ```event``` and one of its ```sessions```. The usual path is: 
1. Choose a ```schedule``` (based on an year);
2. Get an ```Event``` (a race weekends);
3. Get a ```Session``` (individual practice, qualifying, or race sessions);
4. Load the data and navigate ```laps```, ```results```, telemetry, messages, weather, and circuit information;

The public namespaces include ```fastf1.core```, ```fastf1.events```, ```fastf1.ergast```, ```fastf1.livetiming```, ```fastf1.plotting```, ```fastf1.utils```, and ```fastf1.legacy```. 

```fastf1.Cache``` controls local HTTP/data caching and should b enabled once at process startup (should be treated as infrastructure, not as a per-request convenience, that improves iteration speed and reduces unnecessary traffic).

**Limits**<br>
Loading remote data may fail because of connectivity, rate limits, upstream availability, or an unsupported session. 

### Data Sources ###
There're 3 main data sources:
- ```fastf1```: API's own backend (supports from 2018 on; default);
- ```f1timing```: data from the *F1 Live Timing* API (supports from 2018 on; sessions with no timing data are not listed);
- ```ergast```: data from Ergast (supports from 1950 on; no local times available);

### ```Events``` ###
An ```event``` summarizes a racing weekend.

**Event**<br>
Represents a single race weekend or testing event. Each ```event``` contains:
- ```RoundNumber```: round number;
- ```Country```: country where ```event``` takes place;
- ```Location```: circuit name;
- ```EventName```: ```event``` short name;
- ```OfficialEventName```: ```event``` full name
- ```EventDate```: ```event``` date (usually race day);
- ```EventFormat```: format type (```conventional```, ```sprint```, ```sprint_shootout```, ```sprint_qualifying```, ```testing```);
- ```Session[X]```: each session name;
- ```Session[X]Date```: each session local timestamp;
- ```Session[X]DateUtc```: each ssession UTC timestamp;
- ```F1ApiSupport```: whether API data is available;

**Session**<br>
Represents one of the many sessions within an ```event```. Each ```session``` is defined by:
- ```event```: associated event;
- ```session_name```: session name (```Qualifying```, ```Race```, ```FP1```);

Each ```session``` contains:
- ```event```: associated event;
- ```name```: session name;
- ```date```: session timestamp;
​- ```drivers```: list of all driver numbers that participated in session;
- ```results```: session results with driver information;
- ```laps```: all laps from all drivers in session;
- ```car_data```: dictionary of car telemetry (speed, RPM, etc) keyed by driver number (each value is a ```Telemetry``` object);
- ```pos_data```: dictionary of car position data (X, Y, Z coordinates) keyed by driver number (each value is a ```Telemetry``` object);
​- ```weather_data```: DataFrame containing weather data for session;
- ```session_status```: session status data (```Started```, ```Finished```, ```Aborted```);
- ```track_status```: track status data (yellow flags, red flags, etc);
- ```race_control_messages```: race control messages for session;
- ```session_info```: session information including meeting, session, country and circuit names and uuids;
- ```total_laps```: originally scheduled number of laps for race-like sessions;
- ```t0_date```: timestamp marking the beginning of data stream (when session time is zero);
- ```session_start_time```: session time at which the session was started;

**Laps**<br>
```Laps``` class provides multiple laps timing data and telemetry. Each ```Laps``` contains:
- ```Time```: lap completion time;
- ```Driver```: driver 3-letter abbreviation;
- ```DriverNumber```: driver number;
- ```LapTime```: lap time duration
- ```LapNumber```: lap number;
- ```Stint```: stint number;
- ```PitOutTime```: time when car exited pit lane;
- ```PitInTime```: time when car entered pit lane;
- ```Sector[X]Time```: sector [X] time (1-3);
- ```Sector[X]SessionTime```: session time at sector [X] (1-3);
- ```SpeedI[X]```: speed trap at intermediate [X] (km/h; 1-2);
- ```SpeedFL```: speed trap at finish line (km/h);
- ```SpeedST```: speed trap at speed trap (km/h);
- ```IsPersonalBest```: whether lap is driver’s personal best;
- ```Compound```: tyre compound (```SOFT```, ```MEDIUM```, ```HARD```, ```INTERMEDIATE```, ```WET```);
- ```TyreLife```: laps completed on current set of tyres;
- ```FreshTyre```: whether tyres were new;
- ```Team```: team name;
- ```LapStartTime```: time when the lap started;
- ```LapStartDate```: date when the lap started;
- ```TrackStatus```: track status flags during the lap;
- ```Position```: position at end of lap;
- ```Deleted```: whether lap time was deleted;
- ```DeletedReason```: reason for deletion;
- ```FastF1Generated```: whether lap was generated by *FastF1*;
- ```IsAccurate```: whether lap passed accuracy validation;

**Telemetry**<br>
```Telemetry``` contains multiple data channels:
- ```Car```
    - ```Speed```: car speed in km/h;
    - ```RPM```: engine RPMs;
    - ```nGear```: current gear number (1-8);
    - ```Throttle```: throttle pedal position (0-100%);
    - ```Brake```: whether brakes are applied (bool);
    - ```DRS```: DRS (Drag Reduction System) status (```0```: not available, ```1```: available but not activated, ```2```: available and activated, ```3```: active with different detection);
- ```Position```
    - ```X```: X coordinate position in 1/10 meters;
    - ```Y```: Y coordinate position in 1/10 meters;
    - ```Z```: Z coordinate position in 1/10 meters;
    - ```Status```: position status (```OnTrack```, ```OffTrack```);
- ```Timing```
    - ```Time```: relative time to start of the data slice (0: first sample);
    - ```SessionTime```: time elapsed since start of session;
    - ```Date```: date/time the sample was created;
    - ```Source```: how sample was created (```car```: original car sample, ```pos```: original position sample, ```interpolated```: artificially created/interpolated sample);
- ```Computed```
    - ```Distance```: distance driven since first sample (meters);
    - ```RelativeDistance```: relative distance from 0 (first sample) to 1 (last sample);
    - ```DifferentialDistance```: distance driven between consecutive samples (meters);
    - ```DriverAhead```: driver number of car ahead;
    - ```DistanceToDriverAhead```: distance to the car ahead (meters);
